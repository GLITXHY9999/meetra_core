## Dataset Ingestion

````python
import pandas as pd

# Load directly from GitHub URL (runs automatically in Google Colab)
url = 'https://raw.githubusercontent.com/GLITXHY9999/meetra_core/main/employee_data.csv'
df = pd.read_csv(url)

# Alternatively, if uploaded to Colab directly:
# df = pd.read_csv('/content/employee_data.csv')

print(df.head(10).to_string())
````

## Exploratory Data Analysis

### Display Basic Information

````python
df.info()
````

### Summary Statistics

````python
df.describe()
````

### Check for Missing Values

````python
df.isnull().sum()
````

### Correlation Matrix

````python
# Select numeric columns for correlation
df1 = df.select_dtypes(include=['float64', 'int64'])
df1.corr()
````

### Heatmap of Correlations

````python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 8))
sns.heatmap(df1.corr(), annot=False, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
````

### Histograms for Numeric Columns

````python
# Plot histograms for core numeric workforce attributes
key_cols = ['Age', 'MonthlyIncome', 'TotalWorkingYears', 'DistanceFromHome']
df[key_cols].hist(figsize=(8, 6), color='green', edgecolor='black')
plt.suptitle('Histograms of Numeric Columns')
plt.tight_layout()
plt.show()
````

### Boxplot for Outlier Detection

````python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df[['Age', 'TotalWorkingYears', 'DistanceFromHome']])
plt.title('Box Plot for All Numeric Columns')
plt.show()
````

### Scatter Plot

````python
plt.figure(figsize=(8, 5))
plt.scatter(df['Age'], df['MonthlyIncome'], color='purple', alpha=0.7)
plt.title('Scatter Plot: Age vs Monthly Income')
plt.xlabel('Age')
plt.ylabel('Monthly Income ($)')
plt.grid(True)
plt.show()
````

### Pairplot for Feature Relationships

````python
sns.pairplot(df[['Age', 'MonthlyIncome', 'TotalWorkingYears', 'DistanceFromHome']])
plt.suptitle('Pairplot of Dataset', y=1.02)
plt.show()
````

### Visualization 1: Histogram (Feature Distributions)
```python
df[['Age', 'MonthlyIncome']].hist(figsize=(8, 4), color='teal', edgecolor='black')
plt.show()
```
**Observation 1:**
- **Age Distribution:** The employee age distribution is bell-shaped and centered around the **30–38 age bracket**, indicating a primarily mid-career workforce with fewer entry-level (<22) or late-career (>55) personnel.
- **Monthly Income Skew:** The monthly income distribution is **positively skewed (right-skewed)**. Most employees earn between \$2,000 and \$5,000 per month, while a small minority of executive/senior staff earn upwards of \$15,000+.

---
### Visualization 2: Scatter Plot (Age vs. Monthly Income)
```python
plt.scatter(df['Age'], df['MonthlyIncome'], color='purple')
plt.title('Age vs Monthly Income')
plt.xlabel('Age')
plt.ylabel('Monthly Income')
plt.show()
```
**Observation 2:**
- **Positive Correlation:** There is a moderate positive correlation between age and compensation. As age increases, maximum monthly earning potential expands significantly.
- **Bifurcated Compensation Bands:** Above age 40, compensation splits into two distinct clusters: individual contributors remaining in the \$4,000–\$7,000 range, and leadership roles rising above \$15,000.

---
### Visualization 3: Box Plot (Outlier Detection in Distance & Tenure)
```python
sns.boxplot(x=df['DistanceFromHome'], color='coral')
plt.title('Box Plot of Distance From Home')
plt.show()
```
**Observation 3:**
- **Commute Clustering:** The median distance from home is **7 miles**, with 75% of employees living within **14 miles** of the office.
- **Outlier Commuters:** There are several data points extending past 25–29 miles. These extreme commute distances represent high-fatigue employees who are strong candidates for hybrid/remote work accommodations to reduce turnover risk.
