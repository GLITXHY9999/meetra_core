## 5. Algorithmic Fairness & EEOC 80% Four-Fifths Rule Auditor
Inspect regulatory compliance (Title VII / 29 C.F.R. § 1607.4D / ADEA Age cohorts) directly from the model's telemetry.

```python
fairness = engine.metrics.get('fairness_audit', {})
if fairness:
    print('=' * 65)
    print('⚖️ EEOC ALGORITHMIC FAIRNESS & DISPARATE IMPACT AUDIT')
    print('=' * 65)
    print(f'Overall Status       : {fairness.get("overall_status")}')
    print(f'Fairness Score       : {fairness.get("fairness_score")} / 100.0')
    print('-' * 65)
    for cohort, audit in fairness.get('demographics', {}).items():
        air = audit.get('adverse_impact_ratio', 0) * 100
        status = 'PASS' if audit.get('is_compliant') else 'FLAG'
        print(f'Cohort: {cohort.upper():<12} | AIR: {air:5.1f}% | 80% Rule: {status} | Ref: {audit.get("benchmark_group")}')
    print('=' * 65)
else:
    print('Fairness audit telemetry not found.')
```

## 6. TreeSHAP Game-Theoretic Explainability & Team Churn Contagion
Test an individual employee profile to observe exact game-theoretic additive contributions and sociometric turnover contagion.

```python
sample_profile = {
    'Age': 29,
    'Gender': 'Female',
    'Department': 'Sales & Revenue',
    'JobRole': 'Sales Representative',
    'JobLevel': 1,
    'MonthlyIncome': 3200,
    'TotalWorkingYears': 3,
    'YearsAtCompany': 1,
    'YearsWithCurrManager': 0,
    'YearsSinceLastPromotion': 1,
    'JobSatisfaction': 1,
    'EnvironmentSatisfaction': 1,
    'WorkLifeBalance': 1,
    'StockOptionLevel': 0,
    'MaritalStatus': 'Single',
    'BusinessTravel': 'Travel_Frequently',
    'DistanceFromHome': 35,
    'NumCompaniesWorked': 2,
    'Education': 'Bachelors',
    'PerformanceRating': 3,
    'PercentSalaryHike': 12,
    'TrainingTimesLastYear': 1,
    'OverTime': 'Yes'
}

pred = engine.predict_detailed(sample_profile)

print('=' * 60)
print(f'Predicted Flight Risk: {pred["risk_score"]*100:.1f}%')
print(f'Decision Threshold   : {pred["decision_threshold"]*100:.1f}%')
print(f'SHAP Base Value      : {pred.get("shap_base_value")}')
print('-' * 60)
print('Top Marginal Drivers (TreeSHAP):')
for d in pred['drivers'][:5]:
    sign = '+' if d['impact'] > 0 else ''
    print(f'  • {d["feature"]:<8} : {sign}{d["impact"]:.1f} pts ({d["method"]})')
print('-' * 60)
contagion = pred.get('team_contagion', {})
print(f'Team Contagion Tier  : {contagion.get("contagion_risk_level")}')
print(f'Contagion Multiplier : {contagion.get("contagion_multiplier")}x')
print(f'Adjusted Flight Risk : {contagion.get("contagion_adjusted_risk", 0)*100:.1f}%')
print('=' * 60)
```
